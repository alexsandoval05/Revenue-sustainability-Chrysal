## **Simulacion de datos artificiales**

```python
# ============================================================
# Imports y configuración (ejecuta esta celda primero)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 160)

# Semilla global para reproducibilidad
RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

print("Librerías cargadas.")
```

```python
# Variables categóricas
n_rows = 700
rng = np.random.default_rng(RANDOM_SEED)
```

```python
rango_fechas = pd.date_range(start="2023-01-01", end="2025-06-30", freq="D")
sales_order_number = [f"SO-{str(i).zfill(6)}" for i in range(1, n_rows + 1)]
fecha = rng.choice(rango_fechas, size=n_rows)
pais = rng.choice(["Colombia", "Holanda","Kenya","Ecuador"], size=n_rows, p=[.25, .24, .26, .25])
product_reference = rng.choice(["Chrysal Grow X", "Chrysal Root R","Chrysal Boost Z","Chrysal Protect Y","Chrysal Flora F"], size=n_rows, p=[.20, .19, .21, .17, .23])
campaign_id = rng.choice(["1", "2", "3","4"], size=n_rows, p=[.25, .35, .15, .25 ])
category_product = rng.choice(["Fungicida","Insecticida","Regulador de creci","Nutricional","Fertilizante"], size=n_rows, p=[.20, .19, .21, .17, .23])
```

```python
# Ticket: distribución normal que DEPENDE del modelo de pricing
sales_revenue = np.where(
    pais == "Holanda",
    rng.normal(150000, 170000, n_rows),   # holanda: media $150000
    rng.normal(80000, 30000, n_rows)    # colombia: media $120000
)
sales_revenue = np.maximum(sales_revenue, 30000).round(2)  # piso de $5, redondear a centavos
```

### **Ensamblar el dataset incluyendo "sales_revenue" y "production_cost_usd"**

```python
df = pd.DataFrame({
    "sales_order_number": sales_order_number,
    "fecha": fecha,
    "pais": pais,
    "product_reference": product_reference,
    "campaign_id": campaign_id,
    "category_product": category_product,
    "sales_revenue": sales_revenue[:n_rows],
    #"sales_rev": rng.normal(96000, n_rows, size=n_rows).round(1),
    "production_cost_usd": rng.normal(51000, n_rows, size=n_rows).round(1)
})

print(f"Dataset generado: {df.shape[0]} filas × {df.shape[1]} columnas")
```

```python
df.to_csv("Tabla1_Chrysal_sustainability.csv", index=False)
```
